```python
from model.watermark_encoder import WatermarkEncoder
from datasets.celeba import get_celeba_dataloader
```

```python
dataloader = get_celeba_dataloader(
        root="./data/celeba",
        split="train",
        image_size=128,
        batch_size=32,
        num_workers=4,
        shuffle=True,
        return_repr=True,
    )
dataloader = iter(dataloader)
```

```python
batch = next(dataloader)
```

```python
[images, vecs, _] = batch
images.shape
```

```python
from insightface.app import FaceAnalysis

model = FaceAnalysis(name="buffalo_l")
model.prepare(ctx_id=0, det_thresh=0.5, det_size=(640, 640))
```
