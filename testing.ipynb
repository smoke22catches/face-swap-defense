```python
import cv2
import matplotlib.pyplot as plt
```

# Testing watermark recovering from original image

```python
%conda run -n face-swap-defense python test_encoder.py \
--source_image data/celeba/img_align_celeba/000010.jpg \
--image_output_path watermarked_image.jpg \
--message_output_path message.npy \
--message_dim 128 \
--identity_encoder weights/message-recovery-md-opt-watermark-40/identity_encoder/identity_encoder_40.pth \
--message_encoder weights/message-recovery-md-opt-watermark-40/message_encoder/message_encoder_40.pth \
--watermark_encoder weights/message-recovery-md-opt-watermark-40/watermark_encoder/watermark_encoder_40.pth
```

```python
original_image = cv2.imread("data/celeba/img_align_celeba/000010.jpg")
original_image = cv2.cvtColor(original_image, cv2.COLOR_BGR2RGB)
watermarked_image = cv2.imread("watermarked_image.jpg")
watermarked_image = cv2.cvtColor(watermarked_image, cv2.COLOR_BGR2RGB)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 5))
ax1.imshow(original_image)
ax2.imshow(watermarked_image)

ax1.set_title('Original image')
ax2.set_title('Watermarked image')

ax1.axis('off')
ax2.axis('off')

plt.tight_layout()
```

```python
original_image = cv2.imread("data/celeba/img_align_celeba/000010.jpg")
original_image = cv2.cvtColor(original_image, cv2.COLOR_BGR2RGB)
watermarked_image = cv2.imread("watermarked_image.jpg")
watermarked_image = cv2.cvtColor(watermarked_image, cv2.COLOR_BGR2RGB)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 5))
ax1.imshow(original_image)
ax2.imshow(watermarked_image)

ax1.set_title('Original image')
ax2.set_title('Watermarked image')

ax1.axis('off')
ax2.axis('off')

plt.tight_layout()
```

```python
%conda run -n facefusion python ../facefusion/facefusion.py headless-run \
-s watermarked_image.jpg \
-t data/celeba/img_align_celeba/000020.jpg \
-o deepfake_image.jpg
```

```python
watermarked_image = cv2.imread("watermarked_image.jpg")
watermarked_image = cv2.cvtColor(watermarked_image, cv2.COLOR_BGR2RGB)
target_image = cv2.imread("data/celeba/img_align_celeba/000020.jpg")
target_image = cv2.cvtColor(target_image, cv2.COLOR_BGR2RGB)
deepfake_image = cv2.imread("deepfake_image.jpg")
deepfake_image = cv2.cvtColor(deepfake_image, cv2.COLOR_BGR2RGB)

fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(10, 5))
ax1.imshow(watermarked_image)
ax2.imshow(target_image)
ax3.imshow(deepfake_image)

ax1.set_title('Watermarked image')
ax2.set_title('Target image')
ax3.set_title('Deepfake image')

ax1.axis('off')
ax2.axis('off')
ax3.axis('off')

plt.tight_layout()
```
